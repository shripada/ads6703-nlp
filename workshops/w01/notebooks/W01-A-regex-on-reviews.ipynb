## 2 · Regex in plain Python (8 min)

Three functions do most of the work:

| function | returns |
|---|---|
| `re.search(p, s)` | the **first** match (or `None`) |
| `re.findall(p, s)` | a **list** of all matches |
| `re.sub(p, repl, s)` | the string with matches **replaced** |

Always write patterns as **raw strings**: `r"\d+"`, not `"\d+"`.

```python
# Goal: pull every price out of a review. This made-up one has two prices ($24.99 and $80),
# plus a model number (WDT970SAHZ0) whose digits must NOT count as a price
s = "Fits my Whirlpool WDT970SAHZ0 perfectly. Paid $24.99 here, Sears wanted $80 for the same part!"

# \$   a literal dollar sign (a bare $ means "end of text" in regex, so it needs the backslash)
# \d+  one or more digits
print(re.findall(r"\$\d+", s))               # first attempt

# (?:\.\d{2})?  optionally: a literal dot, then exactly two digits (the cents)
print(re.findall(r"\$\d+(?:\.\d{2})?", s))    # allow optional cents
```

The first pattern **under-matches**: it cuts `$24.99` down to `$24`.
`(?: … )?` is an *optional non-capturing group*: "cents, if present".

⚠️ **Gotcha:** if your pattern contains a *capturing* group `( … )`, `findall` returns **only the group**. Try it:

```python
# Same pattern with ( ) instead of (?: ). findall now returns only what the group captured:
# ".99" for $24.99, and an empty string for $80, which has no cents.
print(re.findall(r"\$\d+(\.\d{2})?", s))   # capturing group: surprise!
```

## 3 · Scale up with pandas (8 min)

Every `re` function has a pandas twin that works on a whole column:

| pandas | does |
|---|---|
| `.str.contains(p)` | True/False per row, good for **flags** |
| `.str.findall(p)` | list of matches per row |
| `.str.extract(p)` | capturing groups → **new columns** |
| `.str.replace(p, r, regex=True)` | cleaning |

Reviews contain raw HTML line breaks (`<br />`). Let's count and clean them.

```python
# Some review text still contains HTML line breaks. <br\s*/?> matches <br>, <br/> and <br />:
#   \s*  any amount of whitespace, including none
#   /?   an optional slash
has_br = reviews["text"].str.contains(r"<br\s*/?>")   # one True/False per review: does it contain a match?
print(f"{has_br.sum():,} reviews contain <br> tags")   # summing True/False counts the Trues; :, adds a thousands separator

# New column "clean": the text with every <br> replaced by a space. Later cells use this column.
reviews["clean"] = reviews["text"].str.replace(r"<br\s*/?>", " ", regex=True)   # regex=True: treat the first argument as a pattern
reviews.loc[has_br, ["text", "clean"]].head(3)   # before vs after, for the first 3 reviews that had tags
```

```python
# Do reviews that talk about price rate differently?
mentions_price = reviews["clean"].str.contains(r"\$\d")   # a dollar sign followed by a digit
# Split the reviews into two groups (False = no price, True = price) and show, per group,
# how many reviews there are and their average rating
reviews.groupby(mentions_price)["rating"].agg(["count", "mean"]).round(2)
```

### ✍️ Exercise 2: the noise complaint flag

Flag reviews that mention **noise**, **noisy** or **loud** (in any case) and compare their average rating.

First, run the naive version below and look at what it catches.

```python
# (?i)        ignore case
# \w*loud\w*  "loud" plus any letters attached to it, so we see the whole word (loudly, cloudy, ...)
naive = (reviews["clean"]
    .str.findall(r"(?i)\w*loud\w*")   # per review: a list of every matching word ([] if none)
    .explode()                         # one row per word instead of one list per review
    .dropna()                          # reviews with no match became NaN after explode: drop them
    .str.lower())                      # so "Loud" and "loud" count as the same word
naive.value_counts().head(8)           # the 8 most common matching words
```

😬 Water-filter reviews talk about **cloudy** water. That is a **false positive** for a noise flag.
Fix it with word boundaries `\b` and alternation `(?:a|b|c)`.

```python
noise_pattern = r"..."   # ✍️ your pattern here
reviews["mentions_noise"] = reviews["clean"].str.contains(noise_pattern, case=False)   # case=False: also match Noise, LOUD, ...
reviews.groupby("mentions_noise")["rating"].agg(["count", "mean"]).round(2)            # count and average rating, noise vs no noise
```

```python
# ✅ check
n = reviews["mentions_noise"].sum()   # number of flagged reviews (each True counts as 1)
assert 300 <= n <= 340, f"{n} flagged: expected about 319. Check your word boundaries."
# reviews.mentions_noise is the same column as reviews["mentions_noise"];
# ~ flips True/False, so ~reviews.mentions_noise selects the reviews that do NOT mention noise
print(f"✅ {n} noise reviews, rated {reviews.loc[reviews.mentions_noise, 'rating'].mean():.2f} vs "
      f"{reviews.loc[~reviews.mentions_noise, 'rating'].mean():.2f} overall")
```

## 4 · The manager's question: *how long until it fails?* (10 min)

Customers often write things like *"stopped working after 6 months"*.
If we extract the **number** and the **unit**, we get a rough failure timeline, and nobody had to read 30,000 reviews to get it.

`.str.extract` turns each capturing group into a column:

```python
# Two capturing groups, so .str.extract returns two columns (NaN where there is no match):
#   (\d+)                  the number, e.g. "6"
#   (day|week|month|year)  the unit
#   s?                     an optional plural "s", outside the group so it isn't kept
after = reviews["clean"].str.extract(r"(?i)\bafter\s+(\d+)\s+(day|week|month|year)s?\b")
after.columns = ["n", "unit"]   # name the two columns (extract calls them 0 and 1)
after = after.dropna()          # keep only the reviews that matched
print(f"{len(after)} reviews mention 'after <number> <unit>'")
after.head()
```

```python
# Put every mention on one scale, months: "2 weeks" → 2 × 1/4.3 ≈ 0.47 months
to_months = {"day": 1 / 30, "week": 1 / 4.3, "month": 1, "year": 12}   # months per unit (a month is about 4.3 weeks)
# extract returns text, so astype(float) makes n a number;
# .str.lower() because (?i) also matched "Month" or "YEAR"; .map looks each unit up in to_months
after["months"] = after["n"].astype(float) * after["unit"].str.lower().map(to_months)
after["rating"] = reviews.loc[after.index, "rating"]   # `after` kept the original row labels, so this lines up

# Only the first 3 years (36 months), so a few "after 20 years" mentions don't squash the chart;
# alpha=0.4 makes dots see-through, so darker spots show where many reviews overlap
ax = after.query("months <= 36").plot.scatter(x="months", y="rating", alpha=0.4, figsize=(7, 3.5))
ax.set_title("'after N units' mentions: time vs star rating");   # the ; hides the title's text output, leaving just the chart
```

### Read before you trust it

Look at some of the matches in context. Does every *"after N months"* describe the **product failing**?

```python
# Return the match in review i with `width` characters of text on each side, so we can read what it meant
def in_context(i, pattern=r"(?i)after\s+\d+\s+\w+", width=70):
    text = reviews.at[i, "clean"]   # the cleaned text of the review with row label i
    m = re.search(pattern, text)     # the first match; m.start() and m.end() are its positions in text
    # max(0, ...) stops the slice start going negative when the match is near the beginning
    return "…" + text[max(0, m.start() - width): m.end() + width] + "…"

# 6 random matches; random_state=1 picks the same 6 every time, so everyone sees the same examples
for i in after.sample(6, random_state=1).index:
    print(f"[{reviews.at[i, 'rating']}★]", in_context(i), "\n")   # star rating, then the snippet
```

Some matches are about the **old** appliance (*"my washer died after 12 years, this part fixed it"*), and some are *good* news
(*"after 6 months, still works great"*). A regex finds **a pattern**, but it doesn't understand what the sentence means.
