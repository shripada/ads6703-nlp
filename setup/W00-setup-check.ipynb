```python
import sys, importlib
from pathlib import Path

ok = True
def check(label, cond, fix=""):
    global ok
    print(("✅ " if cond else "❌ ") + label + ("" if cond else f"  →  {fix}"))
    ok = ok and cond

check(f"Python {sys.version.split()[0]}", sys.version_info[:2] in [(3, 11), (3, 12), (3, 13)],
      "use Python 3.11 or 3.12 (see setup guide step 3)")

for pkg in ["pandas", "numpy", "matplotlib", "sklearn", "nltk", "gensim", "torch", "transformers"]:
    try:
        m = importlib.import_module(pkg)
        check(f"{pkg} {getattr(m, '__version__', '')}", True)
    except Exception as e:
        check(pkg, False, f"quit Jupyter, run uv pip install -r requirements.txt, then start it with uv run jupyter lab  ({type(e).__name__})")

data = Path("data") if Path("data").exists() else Path("../data")
for f in ["reviews.csv.gz", "products.csv.gz", "support_inbox.csv"]:
    check(f"data/{f}", (data / f).exists(), "download it from the setup page into ADS6703/data/")

if (data / "reviews.csv.gz").exists():
    import pandas as pd
    n = len(pd.read_csv(data / "reviews.csv.gz"))
    check(f"reviews load: {n:,} rows", n == 30000, "re-download reviews.csv.gz")

print("\nAll set for Workshop 1! 🎉" if ok else "\nFix the ❌ items above, then run again.")
```
