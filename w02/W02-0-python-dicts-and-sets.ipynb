## 1 · Lists, and building one in a single line

A **list** holds items in order, and the same item can appear more than once.

| | |
|---|---|
| `xs[0]` · `xs[-1]` | the first · last item |
| `xs[1:3]` | items 1 and 2 (a *slice*: the end is not included) |
| `len(xs)` | how many items |
| `x in xs` | True if `x` is somewhere in the list (Python checks the items one by one) |
| `xs.count(x)` | how many times `x` appears |
| `xs.append(x)` | add `x` at the end |
| `sorted(xs)` | a new list, in order |

A **list comprehension** builds a new list from an old one in a single line:
`[w.upper() for w in words if len(w) > 3]` reads as "`w.upper()`, for each `w` in `words`, keeping only the ones longer than 3".

```python
print(words[0], words[-1], words[1:3])   # first, last, and items 1-2
print(len(words), "words")               # 8: "great" is counted both times it appears
print("leaks" in words, "leak" in words) # True False: `in` looks for exactly this string
print(words.count("great"))              # 2
print(sorted(words))                     # alphabetical; the original list is unchanged
print([w for w in words if len(w) > 4])  # only the words longer than 4 letters
```

## 2 · Sets: no repeats, and fast membership

A **set** holds each item at most once and has no order. `set(list)` throws the repeats away. Checking `x in s` is
instant even for a million items, because Python doesn't look through them one by one.

Sets also do the three operations a search engine needs:

| | | |
|---|---|---|
| `a & b` | **and**: in both | *intersection* |
| `a \| b` | **or**: in either (or both) | *union* |
| `a - b` | **and not**: in `a` but not in `b` | *difference* |

Run the next cell and compare the two counts. In NLP, every word you read is a **token**, and every *distinct*
word is a **type**. Review 1 has 8 tokens but only 7 types, because "great" appears twice.

```python
r1 = set(reviews[1].split())   # the distinct words of review 1
print(len(reviews[1].split()), "tokens,", len(r1), "types")   # 8 tokens, 7 types
print(r1)                      # a set prints in curly braces, in no particular order

r0 = set(reviews[0].split())
print(r0 & r1)                 # words in both reviews
print(r0 | r1)                 # words in either review
print(r0 - r1)                 # words in review 0 that are not in review 1
```

## 3 · Dicts: look a value up by its key

A **dict** maps **keys** to **values**, like a word to its count, or a word to the reviews it appears in.
Each key appears once; assigning to an existing key replaces its value.

| | |
|---|---|
| `d = {"filter": 2, "ice": 1}` | make one |
| `d["filter"]` | the value for a key (a **KeyError** if the key is missing) |
| `d.get("kettle", 0)` | the value, or 0 if the key is missing |
| `d["kettle"] = 5` | add a key, or replace its value |
| `"ice" in d` | True if `"ice"` is a **key** |
| `len(d)` | how many keys |
| `for k, v in d.items():` | loop over the keys and values together |

```python
counts = {"filter": 2, "ice": 1}
print(counts["filter"])          # 2
print(counts.get("kettle", 0))   # 0: no such key, so we get the default instead of an error
counts["kettle"] = 5             # a new key
counts["ice"] = counts["ice"] + 1   # update an existing key: 1 -> 2
print("ice" in counts, 2 in counts) # True False: `in` checks the keys, not the values
for word, n in counts.items():   # each key with its value
    print(f"{word:>8}: {n}")
```

## 4 · Counter: a dict made for counting

Counting is so common that Python has a ready-made dict for it. `Counter(list)` counts the items in one step, and
`.most_common(n)` gives the `n` most frequent as (item, count) pairs. A Counter returns 0 for a missing key instead of an error.

```python
from collections import Counter   # Counter lives in the standard library, no install needed

c = Counter("the filter leaks after a week the end".split())
print(c)                  # Counter({'the': 2, 'filter': 1, ...})
print(c["the"], c["kettle"])   # 2 0: no KeyError for a missing word
print(c.most_common(2))   # the two most frequent words, as (word, count) pairs
```

## 5 · A dict of lists: which reviews mention each word?

A dict's values can be anything, including lists. Here is the key structure of today's class: for every word,
the list of review numbers that contain it. `enumerate(xs)` gives each item **with its position**, so we know which
review we are in.

`d.setdefault(key, [])` returns the list stored at `key`, after first storing an empty list there if the key is new.
So `d.setdefault(w, []).append(i)` means "add `i` to `w`'s list, creating the list if needed".

```python
for i, review in enumerate(reviews):   # i = 0, 1, 2, 3 alongside each review
    print(i, review)

mentions = {}
for i, review in enumerate(reviews):
    for w in set(review.split()):            # set(): each word once per review, even "great great"
        mentions.setdefault(w, []).append(i) # add review i to this word's list
print(mentions["great"], mentions["filter"]) # [1, 3] [0, 1]
```

## 6 · Sorting by a key

`sorted(xs)` sorts by the items themselves. `sorted(xs, key=f)` sorts by `f(item)` instead: `key=len` sorts by length.
Add `reverse=True` for largest first. `min(xs, key=f)` and `max(xs, key=f)` work the same way.

`lambda w: ...` is a one-line function with no name. `key=lambda w: freq[w]` means "sort each word by its count".

```python
ws = ["water", "ice", "delivery", "a"]
print(sorted(ws))                                     # alphabetical
print(sorted(ws, key=len))                            # shortest first
print(sorted(ws, key=len, reverse=True))              # longest first
print(sorted(ws, key=lambda w: freq[w], reverse=True))  # most frequent in the reviews first
print(max(ws, key=len))                               # the longest word
```

## Cheat sheet

| | |
|---|---|
| **List** | `xs[0]` · `xs[-1]` · `xs[1:3]` · `len(xs)` · `x in xs` · `xs.count(x)` · `xs.append(x)` · `sorted(xs)` |
| **Comprehension** | `[f(x) for x in xs if cond]` · two loops: `[w for r in reviews for w in r.split()]` |
| **Set** | `set(xs)` · `x in s` (fast) · `a & b` and · `a \| b` or · `a - b` and not · `len(s)` |
| **Dict** | `d[k]` · `d.get(k, default)` · `d[k] = v` · `k in d` · `len(d)` · `for k, v in d.items()` |
| **Dict of lists** | `d.setdefault(k, []).append(v)` |
| **Counter** | `Counter(xs)` · `c[x]` (0 if missing) · `c.most_common(n)` |
| **Loop with position** | `for i, x in enumerate(xs)` |
| **Sort by** | `sorted(xs, key=len)` · `key=lambda x: ...` · `reverse=True` · `min` / `max` with `key=` |
